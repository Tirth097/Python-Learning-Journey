```python
# SEABORN
# seaborn.pydata.org

import seaborn as sns
import matplotlib.pyplot as plt
```

```python
sns.set_theme()
```

```python
# sns.get_dataset_names()
```

```python
tips = sns.load_dataset("tips")
print(type(tips))
```

```python
tips.head()
```

```python
sns.relplot(
    data = tips,
    x = "total_bill",
    y = "tip",
    hue = "smoker",
    size = "size",
    style = "smoker"
)
```

```python
x = [1,2,3,4,5,6]
y = [i**2 for i in x]

sns.relplot(
    x = x,
    y = y,
    kind = "line",
)
```

```python
# Scatter Plot
tips = sns.load_dataset("tips")
sns.scatterplot(
    data = tips,
    x = "total_bill",
    y = "tip",
    hue = "time"
)
```

```python
# Line Plot
flights = sns.load_dataset("flights")
sns.lineplot(
    data = flights,
    x = "year",
    y = "passengers"
)
# the blue higliht is confidence interval (95 %)
```

```python
flights.head()
```

```python
# categorical graphs
# 1} BAR GRAPH
sns.load_dataset("tips")
sns.barplot(
    data = tips,
    x = "day",
    y = "tip",
    hue = "sex"
)
```

```python
# 2}  box plots

sns.boxplot(
    data = tips,
    x = "day",
    y = "tip"
)
```

```python
# distributional plot
# 1} historaphs

penguins = sns.load_dataset("penguins")
# penguins.head()

sns.histplot(
    data = penguins,
    x = "body_mass_g",
    bins = 20
)
```
