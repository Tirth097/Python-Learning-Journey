## Looking for Missing Values & Duplicate Records

Before diving into the analysis, it's important to make sure the dataset is complete and free from duplicate records.

- Missing values can affect calculations, visualizations, and model performance, while duplicate records may introduce bias by counting the same customer more than once.
- In this step, we'll verify that the dataset is clean and identify any issues that need to be addressed during the data preparation phase.

```python
# Count missing values in each column
missing_values = df.isnull().sum()

# Create a summary table
missing_summary = pd.DataFrame({
    "Missing Values": missing_values,
    "Missing Percentage (%)": (missing_values / len(df) * 100).round(2)
})

# Display only columns with missing values
missing_summary[missing_summary["Missing Values"] > 0]
```

### Observation

The summary table above highlights any columns containing missing values.

As no rows are returned, the dataset is complete and no immediate action is required before moving to the next stage of analysis.

```python
# Duplicate Records

duplicate_count = df.duplicated().sum()

print(f"Total Duplicate Records : {duplicate_count}")
```

### Observation

A duplicate count of **0** indicates that every customer record is unique.

If duplicate records are found, they will be investigated and removed during the data cleaning phase.

# B. Validating the Dataset Structure

Now that we've explored the dataset, it's time to verify whether its structure aligns with our expectations.

We'll check:

- The total number of rows and columns
- Whether the expected columns are present
- If any columns are missing or unexpected

Performing these checks early helps catch structural issues before they affect downstream analysis.

```python
# Dataset Structure Validation

expected_columns = ["Year","CustomerId","Surname","CreditScore","Geography","Gender","Age","Tenure",
                    "Balance","NumOfProducts","HasCrCard","IsActiveMember","EstimatedSalary","Exited"
]

actual_columns = df.columns.tolist()

missing_columns = set(expected_columns) - set(actual_columns)
extra_columns = set(actual_columns) - set(expected_columns)

print("DATASET STRUCTURE VALIDATION")

print(f"Rows                : {df.shape[0]}")
print(f"Columns             : {df.shape[1]}")
print(f"Expected Columns    : {len(expected_columns)}")
print(f"Available Columns   : {len(actual_columns)}")

print("\nSchema Validation")

print(f"Missing Columns     : {missing_columns if missing_columns else 'None'}")
print(f"Unexpected Columns  : {extra_columns if extra_columns else 'None'}")

if not missing_columns and not extra_columns:
    print("\n Dataset schema validation passed.")
else:
    print("\n Dataset schema validation failed.")
```

# Validating Business-Critical Fields

The reliability of business insights depends on the quality and consistency of the underlying data.

In this section, we validate key business-related variables that play a central role in customer churn analysis.

The validation focuses on:

- Credit card ownership
- Customer activity status
- Number of bank products owned
- Churn labels

These checks help ensure that the data accurately represents real-world customer information before moving on to exploratory analysis.

```python
# Validate Credit Card Ownership
credit_card_distribution = (
    df["HasCrCard"]
      .value_counts()
      .sort_index()
      .rename_axis("Has Credit Card")
      .reset_index(name="Customer Count")
)

credit_card_distribution
```

```python
# Percentage Distribution

credit_card_distribution["Percentage (%)"] = (
    credit_card_distribution["Customer Count"] / len(df) * 100
).round(2)

credit_card_distribution
```

### Observation

The **HasCrCard** feature contains only the expected binary values (**0** and **1**), confirming that the data is logically consistent.

Among the **10,000 customers**, **7,055 (70.55%)** own a bank-issued credit card, while **2,945 (29.45%)** do not. This indicates that credit card ownership is common within the customer base and may serve as a useful variable when analyzing customer behavior and churn patterns in later stages of the project.

---

## Customer Activity Status

The **IsActiveMember** column indicates whether a customer is actively using the bank's services.

Expected values:

- **0** → Inactive Customer
- **1** → Active Customer

Since customer engagement is closely related to retention, validating this feature is essential before investigating churn patterns.

```python
# Validate Customer Activity Status

activity_distribution = ( 
    df["IsActiveMember"].value_counts().sort_index().rename_axis("Active Member").reset_index(name="Customer Count")
)

activity_distribution["Percentage (%)"] = (
    activity_distribution["Customer Count"] / len(df) * 100).round(2)

activity_distribution
```

```python
# Validate Binary Consistency

valid_values = {0, 1}

unique_values = set(df["IsActiveMember"].unique())

if unique_values.issubset(valid_values):
    print("IsActiveMember contains only valid binary values.")
else:
    print("Invalid values found:", unique_values)
```

### Observation

The **IsActiveMember** feature passed the binary consistency check, confirming that it contains only the expected values (**0** and **1**).

Out of **10,000 customers**, **5,151 (51.51%)** are active members, while **4,849 (48.49%)** are inactive. The distribution is relatively balanced, providing a strong foundation for investigating whether customer engagement influences churn behavior in later stages of the analysis.

---

## Product Ownership Validation

The **NumOfProducts** column represents the number of banking products currently used by each customer.

Examples of banking products include savings accounts, current accounts, credit cards, loans, and investment services.

Before using this feature for customer segmentation and churn analysis, it is important to verify that the values are valid and understand how customers are distributed across different product categories.

```python
# Product Ownership Distribution

product_distribution = (
    df["NumOfProducts"]
    .value_counts()
    .sort_index()
    .reset_index(name="Customer Count")
    .rename(columns={"index": "Number of Products"})
)

product_distribution["Percentage (%)"] = (
    product_distribution["Customer Count"] / len(df) * 100
).round(2)

product_distribution
```

```python
# Unique Product Counts
list(map(int, sorted(df["NumOfProducts"].unique())))
```

### Observation

The **NumOfProducts** feature contains four valid product categories (**1**, **2**, **3**, and **4**), with no unexpected or invalid values detected.

Most customers own either **one** or **two** banking products, accounting for **96.74%** of the customer base. Only a small proportion of customers use **three (2.66%)** or **four (0.60%)** products.

This distribution appears realistic and confirms that the feature is suitable for further analysis, particularly when examining the relationship between product ownership and customer churn.

---

## C. Binary Variables Consistency Check

Binary variables represent two possible outcomes and are expected to contain only the values **0** and **1**.

To ensure data integrity, each binary field is validated for consistency before proceeding with further analysis.

The binary variables included in this validation are:

- **HasCrCard**
- **IsActiveMember**
- **Exited**

```python
# Binary Variables Consistency Check

binary_columns = [
    "HasCrCard",
    "IsActiveMember",
    "Exited"
]

validation_results = []

for column in binary_columns:

    unique_values = sorted(df[column].unique())

    validation_results.append({
        "Column": column,
        "Unique Values": unique_values,
        "Validation Status": "Passed" if set(unique_values) == {0, 1} else "Failed"
    })

validation_summary = pd.DataFrame(validation_results)

validation_summary
```

---

## D. Churn Label Validation

The **Exited** column is the target variable for this project and indicates whether a customer has left the bank.

The expected values are:

- **0** → Customer Retained
- **1** → Customer Churned

Before using this variable for further analysis, it is important to verify that the labels are correctly encoded and examine the overall distribution of retained and churned customers.

```python
# Churn Distribution

churn_distribution = (
    df["Exited"]
    .value_counts()
    .sort_index()
    .reset_index(name="Customer Count")
    .rename(columns={"index": "Exited"})
)

churn_distribution["Customer Status"] = [
    "Retained",
    "Churned"
]

churn_distribution["Percentage (%)"] = (
    churn_distribution["Customer Count"] / len(df) * 100
).round(2)

churn_distribution
```

```python
# Overall Churn Rate

churn_rate = (
    df["Exited"].mean() * 100
).round(2)

print(f"Overall Churn Rate: {churn_rate}%")
```

```python
# Validate Churn Labels

sorted(df["Exited"].unique())
```

### Observation

The **Exited** column was successfully validated and contains only the expected binary values (**0** for retained customers and **1** for churned customers), confirming that the target variable is correctly encoded.

Out of **10,000 customers**, **7,963 (79.63%)** were retained, while **2,037 (20.37%)** exited the bank. This indicates that approximately one in every five customers has churned.

The overall churn rate of **20.37%** serves as the baseline performance metric for this project and will be used as a reference when evaluating churn patterns across different customer segments in the subsequent analysis.
